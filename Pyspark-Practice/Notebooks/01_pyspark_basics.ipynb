```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col,rank


window_spec = Window.orderBy(
    col("MonthlyIncome").desc()
)


df.withColumn(
    "salaryRank",
    rank().over(window_spec)).select("employeenumber", "monthlyincome", "salaryRank").show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col,dense_rank


window_spec = Window.orderBy(
    col("MonthlyIncome").desc()
)


ranked_df = df.withColumn(
    "salaryRank",
    dense_rank().over(window_spec))
```

```python
ranked_df.show()
```

```python
ranked_df.select(
    "EmployeeNumber",
    "MonthlyIncome",
    "SalaryRank"
).where(
    col("SalaryRank") == 2
).show()
```

```python
ranked_df.select(
    "employeenumber",
    "monthlyincome",
    "salaryRank"
).where(
    col("salaryRank") <= 20
).show()
```

```python
from pyspark.sql.functions import row_number, col

window_spec = Window.orderBy(col("MonthlyIncome").desc())

df = df.withColumn(
    "row_number",
    row_number().over(window_spec)
)

df.show()
```

```python
window_spec = Window.partitionBy("Department") \
    .orderBy(col("MonthlyIncome").desc())

df = df.withColumn(
    "row_num",
    row_number().over(window_spec)
)

df.select(
    "EmployeeNumber",
    "MonthlyIncome",
    "row_num"
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, lag, lead

window_spec = Window.orderBy(col("MonthlyIncome").desc())

df = df.withColumn(
    "previous_salary",
    lag("MonthlyIncome").over(window_spec)
)

df = df.withColumn(
    "next_salary",
    lead("MonthlyIncome").over(window_spec)
)

df.select(
    "EmployeeNumber",
    "MonthlyIncome",
    "previous_salary",
    "next_salary"
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import avg
from pyspark.sql.functions import round


window_spec = Window.partitionBy("Department")

df.withColumn(
    "avg_salary",
    round(avg("MonthlyIncome").over(window_spec), 0)
).select(
    "EmployeeNumber",
    "Department",
    "MonthlyIncome",
    "avg_salary"
).show()


#round off salary to 2 decimal places
```

```python
window_spec = Window.partitionBy("Department") \
    .orderBy("EmployeeNumber") \
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )

df1 = df.withColumn(
    "running_total",
    sum("MonthlyIncome").over(window_spec)


)
```

```python
df1.select(
    "EmployeeNumber",
    "Department",
    "MonthlyIncome",
    "running_total"
).show(100)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import avg, col

window_spec = Window.partitionBy("Department") \
    .orderBy("EmployeeNumber") \
    .rowsBetween(-2, Window.currentRow)

df = df.withColumn(
    "moving_avg",
    avg("MonthlyIncome").over(window_spec)
)
```
